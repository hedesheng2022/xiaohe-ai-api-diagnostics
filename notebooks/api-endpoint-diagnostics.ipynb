# Xiaohe API endpoint diagnostics

This notebook performs a small, credential-free HTTP check against the public Xiaohe API endpoint. A successful HTTP response proves reachability only; it does not prove that a particular model, route, or authentication scheme is supported.

Related guide: https://ai.xiao-he.top/codex-api.html

```python
from urllib.parse import urlparse
import requests

URL = "https://api.xiao-he.top/"
parsed = urlparse(URL)
assert parsed.scheme in {"http", "https"}
response = requests.get(URL, timeout=8, allow_redirects=False)
print({
    "url": URL,
    "status": response.status_code,
    "ok": response.ok,
    "content_type": response.headers.get("content-type"),
    "location": response.headers.get("location"),
    "body_preview": " ".join(response.text.split())[:240],
})
```

## Interpret the result

- DNS, TCP, TLS, and an HTTP status are reachability evidence.
- They are not evidence that a provider accepts an OpenAI-compatible or Anthropic-compatible request.
- Do not place API keys in this notebook. For configuration examples and failure branches, see https://ai.xiao-he.top/claude-code-api-config.html.
- Product entry points: https://www.xiao-he-ai.com/ and https://api.xiao-he.top/.
